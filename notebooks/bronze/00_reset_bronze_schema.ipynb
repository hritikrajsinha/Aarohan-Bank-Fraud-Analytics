```python
from pyspark.sql import *
files=dbutils.fs.ls("/Volumes/aarohan_bank/bronze/aarohan_landing")
target_catalog="aarohan_bank"
target_schema="bronze"

csv_file_paths_list=[]
for file in files:
    if file.name.lower().endswith(".csv"):
        csv_file_paths_list.append(file.path)
csv_file_paths_list.sort()
print(f"DELETING TABLES FROM THE CATALOG :{target_catalog} AND SCHEMA:{target_schema}")
for file_path in csv_file_paths_list:
    file_name=file_path.split("/")[-1].split(".")[0]
    target_table=f"{target_catalog}.{target_schema}.{file_name}"
    print(f"Deleting Table:{file_name}")
    spark.sql(f"DROP TABLE IF EXISTS {target_table}")
    print(f"Table:{file_name} deleted successfully")
```
